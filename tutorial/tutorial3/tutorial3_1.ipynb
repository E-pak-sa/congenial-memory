```python
%matplotlib inline

import networkx as nx
import matplotlib.pyplot as plt


edges = [
         # (src, dst, weight)
         ('as27', 'as15', 1),
         ('as92', 'as70', 1),
         ('as85', 'as60', 1),
         ('as85', 'as63', 1),
         ('as35', 'as63', 1),
         ('as70', 'as60', 1),
         ('as60', 'as70', 1),
         ('as63', 'as60', 1),
         ('as60', 'as63', 1),
         ('as92', 'as70', 1),
         ('as15', 'as70', 1),
         ('as70', 'as15', 1),
         ('as70', 'as20', 1),
         ('as60', 'as20', 1),
         ('as20', 'as99', 1),
         ('as99', 'as20', 1),
         ('as60', 'as99', 1),
         ('as63', 'as99', 1),
         ('as85', 'as35', 1),
         ('as35', 'as85', 1)
        ]


ases = {
    # 'name':(posX, posY)
    'as27':(0,0),
    'as92':(20,0),
    'as85':(40,0),
    'as35':(60,0),
    'as15':(-10,20),
    'as70':(10,20),
    'as60':(30,20),
    'as63':(50,20),
    'as20':(20,40),
    'as99':(40,40)
}

def plotNetwork(graph):
    # set plot size
    plt.figure(3,figsize=(14,8))

    # draw graph  
    nodz = {k:ases[k] for k in ases if k in graph.nodes()}
    nx.draw_networkx_nodes(graph, pos=nodz, nodelist=nodz.keys(), node_color = 'lightblue', node_size=1500)
    nx.draw_networkx_edges(graph, pos=nodz, nodelist = [], edgelist = graph.edges(), node_size=1500)
    nx.draw_networkx_labels(graph, pos=nodz)
    
    # remove axis and plot
    plt.axis('off')
    plt.show()

# create graph
G = nx.DiGraph()
G.add_weighted_edges_from(edges)
    
# draw network
plotNetwork(G)
```
